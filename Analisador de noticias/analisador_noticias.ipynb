```python
import spacy
```

```python
noticia= """A Fatec de Jaú anunciou na segunda-feira, 12 de maio de 2025, 
a abertura de novas vagas para o curso de Desenvolvimento de Software Multiplataforma. 
O diretor Carlos Almeida e a professora Marina Souza apresentaram o projeto em parceria com a Microsoft. 
Segundo Marina Souza, as inscrições vão até 30 de junho e as aulas começam em agosto de 2025. 
O evento foi realizado em São Paulo e contou com a presença da Universidade de São Paulo."""
```

```python
nlp = spacy.load("pt_core_news_sm")
doc = nlp(noticia)
 
# Componentes do pipeline
print(nlp.pipe_names)
```

```python
# Tokenização, lematização e classe gramatical (primeiros 12 tokens)
for token in doc[:12]:
    print(f"Texto: {token.text:12} | Lemma: {token.lemma_:12} | POS: {token.pos_}")
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
pessoas = []
organizacoes = []
locais = []
 
for ent in doc.ents:
    if ent.label_ == "PER" and ent.text not in pessoas:
        pessoas.append(ent.text)
    if ent.label_ == "ORG" and ent.text not in organizacoes:
        organizacoes.append(ent.text)
    if ent.label_ == "LOC" and ent.text not in locais:
        locais.append(ent.text)
```

```python
meses = ["janeiro", "fevereiro", "março", "abril", "maio", "junho",
         "julho", "agosto", "setembro", "outubro", "novembro", "dezembro"]
 
datas = []
 
for i in range(len(doc)):
    if doc[i].text.lower() in meses:
        inicio = i
        fim = i + 1
 
        # dia antes do mês, ex.: "12 de maio"
        if i >= 2 and doc[i - 1].text == "de" and doc[i - 2].pos_ == "NUM":
            inicio = i - 2
 
        # ano depois do mês, ex.: "maio de 2025"
        if i + 2 < len(doc) and doc[i + 1].text == "de" and doc[i + 2].pos_ == "NUM":
            fim = i + 3
 
        datas.append(doc[inicio:fim].text)
```

```python
print("RESUMO ESTRUTURADO DA NOTÍCIA")
print()
print("Pessoas:", pessoas)
print("Organizações:", organizacoes)
print("Datas:", datas)
print("Locais:", locais)
```
